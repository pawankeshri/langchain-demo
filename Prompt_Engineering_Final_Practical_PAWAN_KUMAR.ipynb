```python
# Step 1: Import required libraries
from google import genai
import getpass

# Step 2: Securely enter your API key
api_key = getpass.getpass("Enter your Gemini API key: ")

# Step 3: Initialize the Gemini client
client = genai.Client(api_key=api_key)

# Step 4: Test the connection with a simple request
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Hello Gemini! Can you explain what prompt engineering is in one sentence?"
)

#print(response.text)

```

```python
def ask_gemini(prompt: str) -> str:
    """
    Sends a prompt to Gemini and returns the response text.
    """
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",   # You can change to gemini-1.5-pro or gemini-3.5-flash other available models
            contents=prompt
        )
        return response.text
    except Exception as e:
        return f"Error: {e}"
```

```python
# write an engineered prompt
engineered_prompt = """
You are an expert teacher.
Context: You are teaching a beginner who has no prior knowledge of technology.
Task: Explain Artificial Intelligence clearly.
Audience: A beginner student.
Constraints: Use simple language, limit to 150 words, and include a real-world example.
Output Format: Use headings, bullet points, and end with 2 reflective questions for the student.
"""

engineered_response = ask_gemini(engineered_prompt)
print(engineered_response)
```

```python
zero_shot_prompt = """
Classify the following customer review into one category:
Categories: Positive, Negative, Neutral
Review: "The delivery was late and the product was damaged."
Return only one classification.
"""

zero_shot_response = ask_gemini(zero_shot_prompt)
print(zero_shot_response)
```

```python
few_shot_prompt = """
You are an AI assistant classifying customer reviews.
Categories: Positive, Negative, Neutral

Examples:
Positive: "The product quality is excellent and delivery was fast."
Positive: "I love this item, works perfectly!"
Negative: "The product stopped working after one day."
Negative: "Terrible service, I will not buy again."
Neutral: "The product is okay, nothing special."

Now classify the following review:
Review: "The packaging was fine, but the product is just average."
Return only one classification.
"""

few_shot_response = ask_gemini(few_shot_prompt)
print(few_shot_response)
```

# Zero-Shot vs Few-Shot Prompting

1. **Did providing examples improve consistency?**  
   Yes. The few-shot prompt gave Gemini clear reference points, so the classification was more consistent and aligned with the intended categories.

2. **Which approach was easier to control?**  
   Few-shot prompting was easier to control because the examples guided Gemini toward the expected style and categories. Zero-shot left more room for ambiguity.

3. **When would Few-Shot Prompting be useful in real applications?**  
   - Customer review classification (like this scenario).  
   - Sentiment analysis for social media posts.  
   - Categorizing support tickets (Billing, Technical, Account Access, Other).  
   - Any task where **consistent labeling** is critical and examples can reduce misclassification.  

✅ **Conclusion:** Few-shot prompting improves reliability and control, especially in classification tasks where categories must be applied consistently.

```python
study_prompt_template = """
Role: You are an expert study planner.
Context: You are helping a student prepare for a subject in a limited number of days.
Task: Generate a personalized study plan.
Audience: Beginner student.
Requirements:
- Include daily topics.
- Suggest practice activities.
- Add a revision schedule.
Constraints:
- Use simple English.
- Keep the plan under 300 words.
- Organize with headings and bullet points.
Output Format:
- Numbered days with topics.
- Bullet points for activities.
- Final section for revision.
Subject: {subject}
Available Days: {days}
Experience Level: {level}
"""
```

```python
def generate_study_plan(subject, days, level):
    prompt = study_prompt_template.format(subject=subject, days=days, level=level)
    return ask_gemini(prompt)

# Example test
plan1 = generate_study_plan("Python Basics", 5, "Beginner")
print(plan1)
```
